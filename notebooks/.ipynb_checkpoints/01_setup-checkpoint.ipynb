```python
from pathlib import Path

PROJECT_ROOT = Path.home() / 'Documents' / 'tcga_hnsc_test'
DATA_DIR = PROJECT_ROOT / 'data'

expr_path = DATA_DIR / 'TCGA-HNSC.star_fpkm-uq.tsv.gz'
print(f'Expression file exists: {expr_path.exists()} {expr_path}')
```

```python
import pandas as pd

expr_path = DATA_DIR / 'TCGA-HNSC.star_fpkm-uq.tsv.gz'
df = pd.read_csv(expr_path, sep='\t', compression='gzip')
print(f'Raw table shape (rows, cols): {df.shape}')
print(f'First columns: {df.columns[:5].tolist()}')
print(f'First gene IDs (top 5): {df.iloc[:5, 0].tolist()}')
df.head(3)
```

```python
gene_col = df.columns[0]

print('=' * 50)
print('EXPRESSION MATRIX INITIALIZATION')
print('=' * 50)

print(f'Gene ID column: {gene_col}\n')

genes = df[gene_col].astype(str)

print(f'Number of genes: {genes.shape[0]}')
print('Example gene IDs (first 10):')
for g in genes.iloc[:10]:
    print(f"  - {g}")

X = df.drop(columns=[gene_col])
print(f'\nRaw numeric matrix shape (genes × samples): {X.shape}')

X.index = genes
X = X.T

print(f'Final expression matrix shape (samples × genes): {X.shape}')
print('=' * 50)
```

```python
import numpy as np

X_log = np.log2(X + 1)

print('=' * 50)
print('LOG-TRANSFORMED EXPRESSION MATRIX')
print('=' * 50)
print(f'Shape (samples × genes): {X_log.shape}\n')

gene_var = X_log.var(axis=0)

print('GENE VARIABILITY SUMMARY')
print('-' * 50)
print(f'Total genes: {gene_var.shape[0]}')
print('Top 10 variance values:')
for v in gene_var.sort_values(ascending=False).head(10):
    print(f'  {v:.3f}')

N_HVG = 2000
top_genes = gene_var.sort_values(ascending=False).head(N_HVG).index
X_hvg = X_log.loc[:, top_genes]

print('\nHIGHLY VARIABLE GENES (HVGs)')
print('-' * 50)
print(f'Selected HVGs: {len(top_genes)}')
print(f'Matrix shape (samples × HVGs): {X_hvg.shape}')

print('\nExample HVGs (first 10):')
for g in top_genes[:10]:
    print(f'  - {g}')

```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_hvg)
print(f'X_scaled shape: {X_scaled.shape}')

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print(f'X_pca shape: {X_pca.shape}')
print(f'Explained variance ratio: {pca.explained_variance_ratio_}')
print(f'Cumulative variance (2 PCs): {pca.explained_variance_ratio_.sum():.3f}')
```

```python
plt.figure(figsize=(7, 6), dpi=140)

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c='red',
    s=22,
    alpha=0.9,
    edgecolor='black',
    linewidth=0.25
)

pc1 = pca.explained_variance_ratio_[0] * 100
pc2 = pca.explained_variance_ratio_[1] * 100

plt.xlabel(f"PC1 ({pc1:.1f}%)")
plt.ylabel(f"PC2 ({pc2:.1f}%)")
plt.title("TCGA-HNSC - PCA (Top 2000 HVGs)", fontsize=12)

plt.tight_layout()

out_path = FIG_DIR / "pca_tcga_hnsc_top2000hvg_red.png"
plt.savefig(out_path, dpi=300, bbox_inches="tight")
print(f"Saved PCA figure to: {out_path}")

plt.show()
```

```python
import umap

reducer = umap.UMAP(
    n_neighbors=15,
    min_dist=0.1,
    n_components=2,
    random_state=42
)

X_umap = reducer.fit_transform(X_scaled)
print(f'X_umap shape: {X_umap.shape}')

plt.figure(figsize=(7, 6), dpi=140)
plt.scatter(
    X_umap[:, 0],
    X_umap[:, 1],
    c="red",
    s=22,
    alpha=0.8,
    edgecolor="black",
    linewidth=0.25
)
plt.xlabel("UMAP1")
plt.ylabel("UMAP2")
plt.title("TCGA-HNSC - UMAP (Top 2000 HVGs)", fontsize=12)
plt.tight_layout()

out_path = FIG_DIR / "umap_tcga_hnsc_top2000hvg_red.png"
plt.savefig(out_path, dpi=300, bbox_inches="tight")
print(f"Saved UMAP figure to: {out_path}")

plt.show()
```

```python
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial.distance import pdist
import numpy as np

dist_matrix = pdist(X_scaled, metric='euclidean')
Z = linkage(dist_matrix, method='ward')
print(f'Linkage matrix shape: {Z.shape}')

k = 3
cluster_labels = fcluster(Z, k, criterion='maxclust')

print(f'Number of clusters: {len(np.unique(cluster_labels))}')
print(f'Cluster sizes: {np.bincount(cluster_labels)[1:]}')
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 6), dpi=140)

scatter = plt.scatter(
    X_umap[:, 0],
    X_umap[:, 1],
    c=cluster_labels,
    cmap='tab10',
    s=25,
    alpha=1.0,
    edgecolor='black',
    linewidth=0.25
)

plt.xlabel('UMAP1')
plt.ylabel('UMAP2')
plt.title('TCGA-HNSC - UMAP with hierarchical clustering', fontsize=12)
plt.legend(
    *scatter.legend_elements(),
    title='Cluster',
    loc='best'
)

plt.tight_layout()

out_path = FIG_DIR / 'umap_tcga_hnsc_clusters_k3.png'
plt.savefig(out_path, dpi=300, bbox_inches='tight')
print(f'Saved clustered UMAP to: {out_path}')

plt.show()
```

```python
from pathlib import Path
import urllib.request

counts_url = 'https://gdc.xenahubs.net/download/TCGA-HNSC.star_counts.tsv.gz'
counts_path = DATA_DIR / "TCGA-HNSC.star_counts.tsv.gz"

print(f'Target path: {counts_path}')

if counts_path.exists():
    print(f'Counts file already exists: {counts_path}')
else:
    print(f'Downloading from: {counts_url}')
    urllib.request.urlretrieve(counts_url, counts_path)
    print(f'Downloaded: {counts_path} (size bytes: {counts_path.stat().st_size})')
```

```python
import pandas as pd

counts_path = DATA_DIR / 'TCGA-HNSC.star_counts.tsv.gz'
counts_df = pd.read_csv(
    counts_path,
    sep='\t',
    compression='gzip'
)

print(f'Counts raw shape (genes x samples+1): {counts_df.shape}')
print(f'First columns: {counts_df.columns[:5].tolist()}')

counts_df.head(3)
```

```python
gene_col_counts = counts_df.columns[0]
print(f'Gene ID column in counts {gene_col_counts}')

genes_counts = counts_df[gene_col_counts].astype(str)

X_counts = counts_df.drop(columns=[gene_col_counts])

print(f'Counts matrix shape (genes x samples): {X_counts.shape}')
```

```python
gene_col_counts = counts_df.columns[0]
genes_counts = counts_df[gene_col_counts].astype(str)

X_counts = counts_df.drop(columns=[gene_col_counts])
X_counts.index = genes_counts
X_counts = X_counts.T           
print('Counts matrix prepared\n')

print(f"Shape (samples × genes): {X_counts.shape}\n")

print("Example sample IDs:")
for s in X_counts.index[:10]:
    print(f"  - {s}")

print("\nExample gene IDs:")
for g in X_counts.columns[:10]:
    print(f"  - {g}")
```

```python
import pandas as pd
import numpy as np

clusters = pd.Series(cluster_labels, index=X_hvg.index, name='cluster')

common_samples = X_counts.index.intersection(clusters.index)
print(f'Common samples: {len(common_samples)}')

X_counts_sync = X_counts.loc[common_samples]
clusters_sync = clusters.loc[common_samples]

print(f'Counts synced shape: {X_counts_sync.shape}')
print(f'Clusters synced shape: {clusters_sync.shape}')
print('Cluster sizes: ')
print(clusters_sync.value_counts().sort_index())
```

```python
lib_size = X_counts_sync.sum(axis=1)

X_cpm = X_counts_sync.div(lib_size, axis=0) * 1e6

X_logcpm = np.log2(X_cpm + 1)

print(f'X_logcpm shape: {X_logcpm.shape}')
```

```python
from scipy.stats import ttest_ind

cluster_a = 1
cluster_b = 2

mask_a = clusters_sync == cluster_a
mask_b = clusters_sync == cluster_b

Xa = X_logcpm.loc[mask_a]
Xb = X_logcpm.loc[mask_b]

print(f'Cluster {cluster_a} samples: {Xa.shape[0]}')
print(f'Cluster {cluster_b} samples: {Xb.shape[0]}')

t_stat, pvals = ttest_ind(Xa.values, Xb.values, axis=0, equal_var=False, nan_policy='omit')

log2fc = Xa.mean(axis=0) - Xb.mean(axis=0)

de = pd.DataFrame({
    'gene': X_logcpm.columns,
    'log2FC': log2fc.values,
    'pval': pvals
}).set_index('gene')

de_clean = (
    de
    .replace([np.inf, -np.inf], np.nan)
    .dropna(subset=['pval'])
)

de_clean.head()
```

```python
from statsmodels.stats.multitest import multipletests

de_clean['fdr'] = multipletests(de_clean['pval'].values, method='fdr_bh')[1]
de_clean['abs_log2FC'] = de_clean['log2FC'].abs()

print(f"NaN FDR: {de_clean['fdr'].isna().sum()}")
```

```python
top10 = (
    de_clean
    .sort_values(['fdr', 'abs_log2FC'], ascending=[True, False])
    .head(10)
)

display(top10)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

top_genes = top10.index.tolist()


heat = X_logcpm.loc[:, top_genes].T

sample_order = clusters_sync.sort_values().index
heat = heat[sample_order]

plt.figure(figsize=(12, 4), dpi=160)

sns.heatmap(
    heat,
    cmap='viridis',
    cbar_kws={'label': 'log2(CPM + 1)'}
)

plt.xlabel('Samples (sorted by cluster)')
plt.ylabel('Top-10 marker genes')
plt.title(
    'Top-10 Differentially Expressed Genes\n'
    'Cluster 1 vs Cluster 2',
    fontsize=12
)

plt.tight_layout()

out_path = FIG_DIR / 'heatmap_top10_cluster1_vs_2.png'
plt.savefig(out_path, dpi=300, bbox_inches='tight')
print(f'Saved heatmap to: {out_path}')

plt.show()

```
